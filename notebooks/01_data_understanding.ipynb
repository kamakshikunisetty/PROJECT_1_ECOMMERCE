## 💳 Payments Table — Phase 1 Understanding

- **Rows:** 103,886
- **Columns:** 5
- **Grain:** One row represents one payment record associated with an order.
- **`order_id`:** Identifies the order, but is not unique because an order can have multiple payment records.
- **`payment_sequential`:** Represents the sequence number of a payment within an order.
- **`payment_installments`:** Represents the number of installments for a payment; it is not the number of payment records.
- **`payment_type`:** Represents the payment method.
- **`payment_value`:** Represents the value of the payment record.
- **Composite key:** `order_id` + `payment_sequential`.
- **Key validation:** The combination of `order_id` and `payment_sequential` has 0 duplicate records, confirming that the combination uniquely identifies a payment record.

### Key takeaway
The Payments table has a **one-to-many relationship with Orders**: one order can have multiple payment records.

# 🔗 Phase 1 — Overall Data Model

## Core Relationships

### Customers → Orders
- `customers.customer_id` → `orders.customer_id`
- Every order customer ID exists in Customers.
- `customer_id` is unique in both tables in this dataset.
- `customer_unique_id` represents the underlying customer identity and is important for repeat-purchase analysis.

### Orders → Order Items
- `orders.order_id` → `order_items.order_id`
- One order can contain multiple order-item records.
- 775 orders are not represented in Order Items and will be investigated during data cleaning.

### Orders → Payments
- `orders.order_id` → `payments.order_id`
- One order can have multiple payment records.
- One order is not represented in Payments and will be investigated later.

### Orders → Reviews
- `orders.order_id` → `reviews.order_id`
- One order can have zero, one, or multiple review records.
- 768 orders have no review record.

### Order Items → Products
- `order_items.product_id` → `products.product_id`
- Every product ID in Order Items exists in Products.
- One product can appear in multiple order-item records.

### Order Items → Sellers
- `order_items.seller_id` → `sellers.seller_id`
- Every seller ID in Order Items exists in Sellers.
- One seller can appear in multiple order-item records.

### Products → Category Translation
- `products.product_category_name` → `category_translation.product_category_name`
- The translation table provides English category names.
- Two product categories do not have translations.

### Customers/Sellers → Geolocation
- `customer_zip_code_prefix` and `seller_zip_code_prefix` can be matched to `geolocation_zip_code_prefix`.
- Geolocation contains multiple records per ZIP prefix.
- Some customer and seller ZIP prefixes have no matching geolocation record.
- Geolocation is therefore treated as a supporting geographic data source rather than a simple one-to-one lookup table.

# 📚 Phase 1 — Data Dictionary

| Table | Grain | Primary/Unique Identifier | Main Purpose |
|---|---|---|---|
| Customers | One row per customer record | `customer_id` | Customer identity and location |
| Orders | One row per order | `order_id` | Order lifecycle and dates |
| Order Items | One row per item within an order | `order_id` + `order_item_id` | Products, sellers, price and freight |
| Payments | One row per payment record | `order_id` + `payment_sequential` | Payment method, installments and value |
| Reviews | One row per review record | `review_id` + `order_id` | Customer rating and written feedback |
| Products | One row per product | `product_id` | Product category and physical attributes |
| Sellers | One row per seller | `seller_id` | Seller identity and location |
| Geolocation | One row per geolocation record | No reliable single unique identifier identified | Geographic coordinates and location |
| Category Translation | One row per category translation | `product_category_name` | Portuguese-to-English category mapping |

## Important Analytical Fields

- `order_purchase_timestamp` → order timing
- `order_delivered_customer_date` → actual delivery
- `order_estimated_delivery_date` → estimated delivery
- `price` → product/item price
- `freight_value` → freight cost
- `review_score` → customer satisfaction
- `review_comment_message` → customer feedback
- `product_category_name` → product category
- `seller_id` → seller performance analysis
- `customer_unique_id` → repeat-customer analysis
