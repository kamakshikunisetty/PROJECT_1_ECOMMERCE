# 🛒 Monthly Average Order Value (AOV) Analysis

## 🎯 Objective

Calculate the Average Order Value (AOV) for each month.

Average Order Value helps us understand:

> How much revenue does the business generate on average from each delivered order?

---

## 📐 AOV Formula

The formula for Average Order Value is:

```text
Average Order Value
=
Total Revenue
÷
Total Number of Orders

````python
monthly_aov_query = """
SELECT
    strftime('%Y-%m', o.order_purchase_timestamp) AS month,

    ROUND(
        SUM(oi.price + oi.freight_value) * 1.0 /
        COUNT(DISTINCT o.order_id),
        2
    ) AS average_order_value,

    COUNT(DISTINCT o.order_id) AS total_orders

FROM orders o

JOIN order_items oi
    ON o.order_id = oi.order_id

WHERE o.order_status = 'delivered'

GROUP BY strftime('%Y-%m', o.order_purchase_timestamp)

ORDER BY month;
"""

monthly_aov = pd.read_sql_query(
    monthly_aov_query,
    conn
)

monthly_aov
````

## 📊 Results and Interpretation

The Average Order Value (AOV) was calculated for each month using:

**AOV = Total Revenue ÷ Number of Unique Orders**

The results show that the Average Order Value generally remained within a relatively stable range throughout the main analysis period.

### Key Observations

- AOV fluctuates across different months rather than continuously increasing.
- Higher total revenue does not necessarily mean customers spend significantly more per order.
- Revenue growth can occur because of:
  1. An increase in the number of orders.
  2. An increase in the average value of each order.
  3. A combination of both.

For example, November 2017 generated the highest total revenue in the analysis. However, its AOV was approximately 158.23.

This indicates that the strong revenue performance during that month was largely associated with a high volume of orders rather than an unusually high Average Order Value.

### Important Note

Months with very few orders, such as September 2016 and December 2016, should be interpreted cautiously because a small number of transactions can produce unstable or unrepresentative AOV values.

Therefore, AOV should be analyzed together with:

- Total Revenue
- Total Number of Orders

This provides a more complete understanding of customer purchasing behavior.

# 📈 Monthly Revenue Growth Analysis

## 🎯 Objective

Analyze how total revenue changes from one month to the next.

This analysis helps identify:

- Revenue growth periods
- Revenue decline periods
- High-growth months
- Changes in business performance over time

---

## 📐 Month-over-Month Growth Formula

Month-over-Month (MoM) growth is calculated using:

**MoM Growth (%) = ((Current Month Revenue - Previous Month Revenue) / Previous Month Revenue) × 100**

A positive value indicates revenue growth.

A negative value indicates a decline in revenue.

---

## 🧠 SQL Concept: Window Functions

A window function performs calculations across related rows without combining them into a single row.

For this analysis, the `LAG()` function is used.

```sql
LAG(total_revenue) OVER (ORDER BY month)

````python
monthly_growth_query = """
WITH monthly_orders AS (
    SELECT
        strftime('%Y-%m', order_purchase_timestamp) AS month,
        COUNT(DISTINCT order_id) AS total_orders
    FROM orders
    WHERE order_status = 'delivered'
    GROUP BY strftime('%Y-%m', order_purchase_timestamp)
)

SELECT
    month,
    total_orders,
    LAG(total_orders) OVER (ORDER BY month) AS previous_month_orders,
    ROUND(
        (
            total_orders - LAG(total_orders) OVER (ORDER BY month)
        ) * 100.0
        / LAG(total_orders) OVER (ORDER BY month),
        2
    ) AS order_growth_percentage
FROM monthly_orders
ORDER BY month;
"""

monthly_growth = pd.read_sql_query(
    monthly_growth_query,
    conn
)

monthly_growth
````

## Repeat Customer Analysis

### Business Question

Which customers have placed more than one order?

### SQL Query

SELECT
    customer_unique_id,
    COUNT(DISTINCT o.order_id) AS total_orders

FROM customers c

JOIN orders o
    ON c.customer_id = o.customer_id

GROUP BY customer_unique_id

HAVING COUNT(DISTINCT o.order_id) > 1

ORDER BY total_orders DESC;


### SQL Concepts

#### GROUP BY

GROUP BY customer_unique_id

Groups all orders belonging to the same actual customer.

---

#### COUNT(DISTINCT)

COUNT(DISTINCT o.order_id)

Counts the number of unique orders placed by each customer.

DISTINCT ensures that an order is counted only once.

---

#### HAVING

HAVING COUNT(DISTINCT o.order_id) > 1

Filters grouped results.

Only customers with more than one order are included.

---

### WHERE vs HAVING

WHERE:
Filters individual rows before grouping.

HAVING:
Filters groups after aggregation.

Example:

WHERE o.order_status = 'delivered'

Filters individual orders.

HAVING COUNT(o.order_id) > 1

Filters customer groups based on their number of orders.


### Key Insight

The dataset contains 2,997 customers who placed more than one order.

The highest-frequency customer placed 17 orders.

````python
repeat_customer_rate_query = """
WITH customer_orders AS (

    SELECT
        c.customer_unique_id,
        COUNT(DISTINCT o.order_id) AS total_orders

    FROM customers c

    JOIN orders o
        ON c.customer_id = o.customer_id

    GROUP BY c.customer_unique_id
)

SELECT
    COUNT(*) AS total_unique_customers,

    SUM(
        CASE
            WHEN total_orders > 1 THEN 1
            ELSE 0
        END
    ) AS repeat_customers,

    ROUND(
        SUM(
            CASE
                WHEN total_orders > 1 THEN 1
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS repeat_customer_rate

FROM customer_orders;
"""

repeat_customer_rate = pd.read_sql_query(
    repeat_customer_rate_query,
    conn
)

repeat_customer_rate
````

    ) AS repeat_customer_rate

### SQL Concepts

#### SUM()

SUM() adds all the values.

Since repeat customers receive a value of 1, adding them gives the total number of repeat customers.


#### COUNT(*)

COUNT(*) counts all rows.

Since the CTE contains one row per unique customer, COUNT(*) gives the total number of unique customers.

Groups all payment records according to their payment method.

Examples:

- credit_card

- voucher
- debit_card


#### COUNT(*)

COUNT(*)

Counts the total number of payment records.

#### COUNT(DISTINCT order_id)

COUNT(DISTINCT order_id)

Counts the number of unique orders associated with each payment type.

DISTINCT ensures that the same order is counted only once.


#### SUM()

SUM(payment_value)

Adds all payment values within each payment type.

This measures the total payment value processed through each method.


#### ROUND()

ROUND(SUM(payment_value), 2)

Rounds the total payment value to two decimal places.

            THEN 1
            ELSE 0

            CASE

FROM orders

WHERE order_status = 'delivered'
AND order_delivered_customer_date IS NOT NULL
AND order_estimated_delivery_date IS NOT NULL;


### SQL Concepts

#### CASE WHEN

CASE WHEN provides conditional logic in SQL.

Syntax:

CASE
    WHEN condition THEN value
    ELSE value
END

In this analysis:

If actual delivery date > estimated delivery date:
    Late = 1

Otherwise:
    Late = 0


#### SUM(CASE WHEN ...)

SUM(
    CASE
        WHEN condition THEN 1
        ELSE 0
    END
)

Counts the number of rows satisfying the condition.

Late orders receive 1.
On-time orders receive 0.

SUM() adds all the 1s.


#### COUNT(*)

COUNT(*) counts all rows that remain after the WHERE

Here, it represents the number of valid delivered orders.


#### ROUND()

ROUND(value, 2)

Rounds a numerical value to 2 decimal places.


#### 100.0

Using 100.0 ensures decimal arithmetic when calculating

### Formula

Late Delivery Rate =

(Late Orders / Valid Delivered Orders) × 100

- Late delivery rate: 8.11%

## 17.4 First Delivery Experience vs Future Repeat Purchase

### Business Question
Are customers who experience a late first delivery less likely to purchase again AFTER receiving that order?

### Why This Analysis Was Improved

A customer might place a second order before receiving their first order.

Example:

Order 1 placed → Jan 1
Order 2 placed → Jan 5
Order 1 delivered late → Jan 20

Therefore, simply checking:

Total Orders > 1

does NOT prove that the customer repeated AFTER experiencing the delivery.

The improved analysis checks:

Another Order Purchase Timestamp
>
First Order Delivery Date

---

### Key SQL Concepts

- CTE (WITH)
- ROW_NUMBER()
- PARTITION BY
- LEFT JOIN
- CASE WHEN
- Date comparison
- GROUP BY

---

### Logic

Step 1:
Find each customer's first valid delivered order.

Step 2:
Classify the first delivery:

Late:
Actual Delivery Date > Estimated Delivery Date

On Time:
Actual Delivery Date <= Estimated Delivery Date

Step 3:
Check whether the customer placed another order AFTER the first order was delivered.

---

### Why LEFT JOIN?

LEFT JOIN keeps all customers, including customers who never placed another order.

If INNER JOIN was used:
Customers with no later orders would disappear.

That would produce an incorrect repeat rate.

Customers with an on-time first delivery had a higher post-delivery repeat rate than customers with a late first delivery.

````python
delivery_review_query = """
SELECT
    CASE
        WHEN o.order_delivered_customer_date
             > o.order_estimated_delivery_date
        THEN 'Late'

        ELSE 'On Time or Early'
    END AS delivery_status,

    COUNT(DISTINCT o.order_id) AS total_orders,

    COUNT(r.review_id) AS total_reviews,

    ROUND(AVG(r.review_score), 2) AS average_review_score

FROM orders o

JOIN reviews r
    ON o.order_id = r.order_id

WHERE o.order_status = 'delivered'
AND o.order_delivered_customer_date IS NOT NULL
AND o.order_estimated_delivery_date IS NOT NULL
AND r.review_score IS NOT NULL

GROUP BY delivery_status

ORDER BY average_review_score DESC;
"""

delivery_review = pd.read_sql_query(
    delivery_review_query,
    conn
)

delivery_review
````
