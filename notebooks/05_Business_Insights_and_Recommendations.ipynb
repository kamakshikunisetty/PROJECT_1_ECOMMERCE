## Project Phase 5: Business Insights

This notebook translates the results of the SQL-based exploratory and business analysis into meaningful business insights and evidence-based recommendations.

The previous analysis examined:

- Revenue trends and month-over-month performance
- Product category revenue concentration and Pareto analysis
- Customer purchasing behavior using an adapted RFM approach
- Delivery delay severity and customer review scores
- Seller-associated operational performance indicators

The objective of this phase is to move beyond reporting metrics and answer the following business questions:

1. What are the most important patterns identified in the data?
2. Why do these patterns matter from a business perspective?
3. What potential actions could the business consider?
4. What limitations should be considered when interpreting the findings?

All insights and recommendations in this notebook are based on the results obtained during the previous analysis phase. Recommendations are intended as data-informed business considerations and do not establish causal relationships unless supported by the available data.

### Key Business Insights

The monthly revenue analysis indicates substantial growth in the platform's product revenue between 2017 and early 2018.

Several important patterns can be observed:

1. **Strong revenue expansion during 2017**
   
   Product revenue increased from approximately 120,313 in January 2017 to over 1 million in November 2017. This indicates significant expansion in transaction activity during this period.

2. **November 2017 recorded the highest complete-month revenue**
   
   Product revenue reached approximately 1.01 million in November 2017, representing the highest revenue level among the complete months in the analysis. However, the available data does not independently establish the specific business factors responsible for this increase.

3. **Revenue remained at a relatively high level during 2018**
   
   During the first eight months of 2018, monthly product revenue generally remained between approximately 845,000 and 997,000. This suggests that the platform maintained a substantially higher revenue level compared with the earlier stages of the dataset.

4. **Revenue volatility remained present**
   
   Although the overall revenue level increased, month-to-month fluctuations continued. For example, revenue declined after the November 2017 peak and experienced additional fluctuations throughout 2018.

5. **Early and final dataset periods should be interpreted cautiously**
   
   Extreme month-over-month percentages in the earliest months are influenced by very small revenue values. Similarly, the very low revenue recorded in September and October 2018 should not be interpreted as evidence of a business decline because these periods may not contain complete monthly data.

### Business Consideration

The results suggest that revenue growth should be evaluated using sustained revenue patterns rather than individual month-over-month percentage changes. Monitoring revenue trends across complete reporting periods would provide a more reliable basis for identifying periods of sustained growth or decline.

## 5.2 Delivery Performance and Customer Satisfaction Insights

The delivery and review analysis identifies a strong association between delivery performance and customer satisfaction.

### Key Business Insights

**1. On-time or early deliveries are associated with substantially higher customer satisfaction**

Orders delivered on time or earlier had an average review score of **4.29**, indicating relatively high customer satisfaction among this group.

**2. Even relatively short delivery delays are associated with a substantial decline in review scores**

Orders delivered **1–7 days late** had an average review score of **2.72**, compared with **4.29** for orders delivered on time or early. This indicates a substantial difference in customer satisfaction between the two groups.

**3. Delays exceeding seven days are associated with very low review scores**

Orders delayed **8–14 days** had an average review score of **1.67**, while orders delayed **15 or more days** had an average score of **1.73**.

Both groups had substantially lower review scores than orders delivered on time or early.

**4. The relationship is not perfectly linear across the most severe delay categories**

The average review score for orders delayed **15 or more days** was slightly higher than that for orders delayed **8–14 days**.

Therefore, the results should not be interpreted as evidence that customer satisfaction continuously decreases with every additional day of delay.

**5. Delivery performance represents an important operational area for further investigation**

The substantial difference in average review scores between on-time and late deliveries indicates that delivery performance is strongly associated with customer experience and should be monitored as an important operational indicator.

### Business Consideration

The business could prioritize further investigation into the operational factors associated with late deliveries, particularly for orders experiencing delays beyond seven days. Potential areas for further investigation may include **logistics processes, delivery regions, and seller-associated order patterns**.

### Important Limitation

This analysis identifies an **association** between delivery delay severity and customer review scores. It does not establish that delivery delays directly caused lower customer reviews, as other factors may also influence customer satisfaction.

## 5.3 Customer Segmentation and Repeat Purchase Insights

The adapted RFM analysis and customer purchase frequency analysis provide insights into customer purchasing patterns, spending behavior, and repeat purchase activity.

### Key Business Insights

**1. Most customers in the dataset made only one purchase**

Approximately **90,557 customers (97.00%)** made only one purchase, while approximately **2.76%** made two purchases. Customers with more than two purchases represented only a small proportion of the analyzed customer population.

This indicates that the observed customer base is predominantly composed of one-time purchasers within the available dataset period.

**2. One-time customers account for a substantial share of customer activity across all spending segments**

The customer segmentation analysis shows that each of the four recency and spending groups contains substantially more one-time customers than repeat customers.

For example, the **More Recent - Higher Spending** segment contained **22,289 one-time customers**, compared with **1,310 repeat customers**.

This highlights the importance of understanding opportunities for encouraging additional purchases among customers who have already completed an initial transaction.

**3. Repeat customers have higher average customer values within comparable spending segments**

Among the higher-spending segments, repeat customers had higher average customer values than one-time customers.

For example:

- **More Recent - Higher Spending repeat customers:** average customer value of **300.23**
- **More Recent - Higher Spending one-time customers:** average customer value of **231.63**

A similar pattern is observed among the less recent higher-spending groups.

This suggests that customers who made repeat purchases generated higher total value per customer within the analyzed dataset.

**4. Higher-spending customer segments contribute substantially more revenue than lower-spending segments**

The higher-spending one-time customer segments generated more than **5 million** in product revenue each, while the corresponding lower-spending segments generated approximately **1.1 million** each.

This indicates that differences in customer spending behavior represent an important source of variation in overall revenue.

**5. Customer recency provides an additional perspective on purchasing activity**

The segmentation analysis distinguishes between more recent and less recent customers based on their purchasing activity relative to the latest available purchase date in the dataset.

This allows customer groups to be viewed using both spending behavior and purchase recency rather than relying only on total revenue.

### Business Consideration

The results suggest that the business could further investigate opportunities to encourage additional purchases among customers who have already completed an initial transaction, particularly among higher-spending customer groups.

Customer groups with higher observed spending may represent useful populations for further analysis of repeat purchase behavior and customer engagement. However, additional data would be required to determine which specific business actions would most effectively increase repeat purchases.

### Important Methodological Consideration

The customer segmentation uses an **adapted RFM approach** based on purchase recency, purchase frequency, and product spending. The segment labels are relative to the analyzed dataset and should not be interpreted as universal customer classifications.

Additionally, the dataset covers a limited historical period. Therefore, a customer classified as a one-time customer in the dataset should not necessarily be interpreted as a customer who never made another purchase outside the available observation period.

## 5.4 Revenue Concentration and Product Category Insights

The revenue concentration analysis examines how categorized product revenue is distributed across product categories.

### Key Business Insights

**1. Product revenue is concentrated among a relatively small proportion of categories**

The analysis included **73 valid product categories**. The top **17 categories**, representing approximately **23.3% of the analyzed categories**, generated more than **80% of categorized product revenue**.

This indicates that a relatively small group of product categories contributes a substantial proportion of the revenue included in the category-level analysis.

**2. The largest individual category does not dominate revenue on its own**

The highest-revenue category, **health_beauty**, contributed approximately **9.38%** of categorized product revenue.

This indicates that revenue concentration exists across a group of high-performing categories rather than being driven by a single category.

**3. The top five categories represent a substantial share of categorized revenue**

The top five product categories together contributed approximately **40.27% of categorized product revenue**.

This suggests that the performance of the leading product categories is important to overall revenue performance.

**4. The revenue distribution does not represent an exact 80/20 Pareto relationship**

Approximately **23.3% of product categories** were required to generate more than 80% of categorized revenue.

Therefore, the results demonstrate substantial revenue concentration, but they should not be described as an exact **80/20 Pareto distribution**.

**5. High-revenue categories may represent strategically important areas for further analysis**

Because a relatively small group of categories contributes a substantial proportion of categorized revenue, changes in the performance of these categories could have a meaningful association with overall revenue patterns.

These categories may therefore represent useful areas for further investigation regarding product demand, pricing, inventory, and customer purchasing behavior.

### Business Consideration

The business could monitor the performance of its highest-revenue product categories while also examining opportunities for growth among lower-revenue categories.

However, revenue concentration alone should not be interpreted as evidence that lower-revenue categories are unimportant. Additional analysis would be required to evaluate factors such as profitability, growth potential, product demand, and strategic importance.

### Important Methodological Consideration

This analysis is based on **categorized product revenue** calculated using product prices. Products with missing category information were excluded from the category-level analysis because they could not be assigned to a meaningful product category.

Therefore, the revenue percentages and concentration measures apply specifically to the analyzed categorized product revenue.

## 5.5 Seller-Associated Operational Performance Insights

The seller-associated operational analysis identifies relative differences in delivery and customer satisfaction indicators among sellers with at least 50 delivered orders.

### Key Business Insights

**1. Most analyzed sellers did not fall into the identified relative concern categories**

Out of **425 analyzed sellers**, **266 sellers (62.59%)** were classified as having **No Relative Concern Identified** based on the dataset-relative delivery and review benchmarks.

This indicates that most sellers in the analyzed population did not simultaneously exhibit comparatively high late-delivery rates or comparatively low review scores.

**2. A smaller group of sellers showed combined delivery and satisfaction concerns**

**52 sellers (12.24%)** were classified under **Combined Delivery & Satisfaction Concern**.

This group had an average late-delivery rate of **13.66%** and an average review score of **3.69**, compared with **4.41%** and **4.28**, respectively, for sellers with no relative concern identified.

These sellers may represent useful candidates for further operational investigation.

**3. Delivery and customer satisfaction indicators do not always identify the same sellers**

Some sellers were classified only under **Higher Delivery Concern**, while others were classified only under **Lower Satisfaction Concern**.

This suggests that delivery performance and customer satisfaction should be evaluated as related but distinct operational indicators.

### Business Consideration

Dataset-relative performance screening can help identify seller-associated order patterns that may require further investigation. However, sellers identified in a relative concern category should not automatically be interpreted as poor-performing sellers.

### Important Limitation

Delivery and review outcomes are recorded at the order level, and an order may contain products from multiple sellers.

Therefore, this analysis identifies **seller-associated patterns** and does not establish that a specific seller directly caused a delivery delay or a particular customer review outcome.

Additionally, the performance categories are based on **dataset-relative percentile benchmarks** and should not be interpreted as universal business performance ratings.
